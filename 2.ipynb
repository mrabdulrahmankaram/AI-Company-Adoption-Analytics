```python
import pandas as pd
import numpy as np

df = pd.read_csv('ai_industry_summary_dirty.csv')

df.head()
```

```python
df.info()
```

```python
df.isnull().sum()
```

```python
df['avg_productivity_change_percent'] = df['avg_productivity_change_percent'].fillna(df['avg_productivity_change_percent'].median())

df['avg_jobs_created'] = df['avg_jobs_created'].fillna(df['avg_jobs_created'].median())
```

```python
df.isnull().sum()
```

```python
df.duplicated().sum()
```

```python
df=df.drop_duplicates()
```

```python
df.duplicated().sum()
```

```python
df['avg_ai_adoption_rate'].unique()
```

```python
df['avg_ai_adoption_rate'] = df['avg_ai_adoption_rate'].str.replace('%', '')
```

```python
df['avg_ai_adoption_rate'] = pd.to_numeric(df['avg_ai_adoption_rate'])
```

```python
df['avg_ai_adoption_rate']
```

```python
df['avg_customer_satisfaction'].unique()
```

```python
df['avg_customer_satisfaction'] = df['avg_customer_satisfaction'].str.replace(',', '.')
```

```python
df['avg_customer_satisfaction'] = pd.to_numeric(df['avg_customer_satisfaction'])
```

```python
df['avg_customer_satisfaction']
```

```python
df['industry'].unique()
```

```python
df['industry'] = df['industry'].str.strip()
```

```python
df['industry'].unique()
```

```python
Q1 = df.select_dtypes(include='number').quantile(0.25)
Q2 = df.select_dtypes(include='number').quantile(0.50)
Q3 = df.select_dtypes(include='number').quantile(0.75)

print("Q1:")
print(Q1)

print("Q2:")
print(Q2)

print("Q3:")
print(Q3)
```

```python
IQR = Q3 - Q1

lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR

print("Lower:")
print(lower)

print("Upper:")
print(upper)
```

```python
numeric = df.select_dtypes(include='number')

outliers = (numeric < lower) | (numeric > upper)

outliers.sum()
```

```python
for col in numeric.columns:
    print(col)
    print(df.loc[outliers[col], col].values)
```
